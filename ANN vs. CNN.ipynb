# **Train-Test Split**

```python
X_train = df.drop("label", axis = 1)
y_train = df["label"]
X_test = df_test
```

```python
X_train = X_train.astype("float32")/255.0
X_test = X_test.astype("float32")/255.0
```

```python
X_train_img = X_train.values.reshape(-1, 28, 28)
X_test_img = X_test.values.reshape(-1, 28, 28)
```

```python
y_train_categ = to_categorical(y_train, 10)
```

# **Model Implementation**

## **1- Perceptron**

```python
perceptron = Sequential(
    [
        Flatten(input_shape=(28,28)),
        Dense(10, activation = "softmax")
    ]
)
```

```python
perceptron.compile(optimizer = "sgd", loss = "categorical_crossentropy", metrics=["accuracy"] )
```

```python
history_percep = perceptron.fit(X_train_img, y_train_categ, epochs = 5, batch_size = 32, verbose = 1)
```

```python
acc_percep = history_percep.history["accuracy"][-1]
```

```python
acc_percep
```

**2- ANN**

```python
ann = Sequential([
    Flatten(input_shape=(28, 28, 1)),

    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
ann.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history_ann = ann.fit(
    X_train_img,
    y_train_categ,
    epochs=5,
    batch_size=32,
    verbose=1
)
```

```python
acc_ANN = history_ann.history["accuracy"][-1]
```

```python
acc_ANN
```

**3- CNN**

```python
cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
cnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history_cnn = cnn.fit(
    X_train_img,
    y_train_categ,
    epochs=5,
    batch_size=32,
    verbose=1
)
```

```python
acc_CNN = history_cnn.history["accuracy"][-1]
```
